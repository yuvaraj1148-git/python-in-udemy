# Comparison operators
x = 10
y = 20
z = x == y  # Equal to
w = x != y  # Not equal to
v = x < y   # Less than
u = x > y   # Greater than
t = x <= y  # Less than or equal to
s = x >= y  # Greater than or equal to

# Logical operators
p = True
q = False
r = p and q  # AND
s = p or q   # OR
t = not p    # NOT

# Assignment operators
m = 10
m += 5   # Equivalent to m = m + 5
m -= 3   # Equivalent to m = m - 3
m *= 2   # Equivalent to m = m * 2
m /= 4   # Equivalent to m = m / 4
m %= 3   # Equivalent to m = m % 3
m **= 2  # Equivalent to m = m ** 2

# Bitwise operators
n = 5  # Binary: 0101
o = 3  # Binary: 0011
p = n & o  # AND
q = n | o  # OR
r = n ^ o  # XOR
s = ~n     # NOT
t = n << 1 # Left shift
u = n >> 1 # Right shift

# Membership operators
list1 = [1, 2, 3, 4, 5]
in_operator = 3 in list1  # True
not_in_operator = 6 not in list1  # True

# Identity operators
x1 = [1, 2, 3]
x2 = [1, 2, 3]
x3 = x1

is_operator = x1 is x2  # False
is_not_operator = x1 is not x2  # True
