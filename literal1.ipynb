# integer literals
# in Python, you can use underscores in numeric literals for better readability. For example, the number 123345678 can be written as 123_345_678. This does not affect the value of the number; it is purely for visual clarity.

```python
x = 25
y = 123_345_678
print(x,y)
```

# float literals
# in Python, float literals are used to represent real numbers. They can be written in decimal form or in scientific notation.
# we can also use underscores in float literals for better readability, just like with integer literals. For example, the number 3.14159 can be written as 3.14_159.

```python
x = 3.14_159
y = 2.5e-3  # scientific notation
a = 1.0e+6  # scientific notation
z = 12.3
print(x, y, z, a)
```

# boolean literals
# in Python, boolean literals are used to represent truth values. There are only two boolean literals: True and False. They are case-sensitive, so they must be written with an uppercase 'T' and 'F'.

```python
x = True
y = False
print(x, y)
```

# complex literals
# in Python, complex literals are used to represent complex numbers. A complex number consists of a real part and an imaginary part. The imaginary part is denoted by a 'j' or 'J' suffix. For example, the complex number 3 + 4j has a real part of 3 and an imaginary part of 4.
# in complex literals, you can also use underscores for better readability, just like with integer and float literals. For example, the complex number 1_000 + 2_000j can be written as 1_000 + 2_000j.

```python
x = 3 + 4j
y = 1_000 + 2_000j
print(x, y)
```

# string literals
# in Python, string literals are used to represent text data. They can be written in single quotes, double quotes, or triple quotes.
# Triple quotes are used for multi-line strings. You can also use escape sequences to include special characters in strings, such as newlines (\n) or tabs (\t).
# if you want to include a single quote in a string that is enclosed in single quotes, you can use the backslash (\) to escape it. For example, 'It\'s a beautiful day!' is a valid string literal.

```python
x = "Hello, World!"
y = 'Python is great!'
z = """This is a multi-line string."""
print(x, y, z)
```

# special literals
# in Python, there are special literals that represent the absence of a value or a null value. The most common special literal is None, which is used to indicate that a variable has no value or that a function does not return anything. It is often used as a placeholder or default value.

```python
x = None
print(x)
```

# escape sequences
# in Python, escape sequences are used to represent special characters in strings. They are preceded by a backslash (\) and allow you to include characters that are otherwise difficult to represent in a string. Some common escape sequences include:
# \n - Newline
# \t - Tab
# \r - Carriage Return
# \\ - Backslash
# \' - Single Quote
# \" - Double Quote
# \ - Backslash
# use escape sequences to format strings or include special characters. For example, you can use \n to create a new line in a string, or \t to add a tab space.

```python
x = "Hello,\nWorld!"
y = "This is a tab:\tSee?"
z = "This is a backslash: \\"
print(x)
print(y)
print(z)
```
