```python
# Nested if 
# example for nested if

temp = 30

if temp == 25:
    print('Temperature is 25')
else:
    if temp < 25:
        print('Temperature is less than 25')
    else:
        print('Temperature is greater than 25')
```

```python
# elif 


temp = 30

if temp == 25:
    print('Temperature is 25')
elif temp == 0:
    print('very cold')
else:
    if temp < 25:
        print('Temperature is less than 25')
    else:
        print('Temperature is greater than 25')
```

```python
# if and elif ladder
temp = 30

if temp == 25:
    print('Temperature is 25')
elif temp == 0:
    print('very cold')
elif temp < 25:
    print('Temperature is less than 25')
elif temp > 25:
    print('Temperature is greater than 25')
else:
    print('Temperature is normal)
```
