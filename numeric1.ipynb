# using sys you can find the size of an object in bytes

```python
import sys

x = 10
y = 12345678912345
z = -12.765778222999

print(sys.getsizeof(x))
print(sys.getsizeof(y))
print(sys.getsizeof(z))
```

# binary representation of a number
# The bin() function is used to convert an integer number to a binary string prefixed with "0b". the bin() function takes an integer as an argument and returns a string representation of the number in binary format.
# if you manually convert a decimal number to binary, first you have to divide the number by 2 and write down the remainder. then you have to divide the quotient by 2 and write down the remainder. you have to repeat this process until the quotient becomes 0. then you have to write down the remainders in reverse order. this is how you can convert a decimal number to binary manually.


# octal representation of a number
# The oct() function is used to convert an integer number to an octal string prefixed with "0o". the oct() function takes an integer as an argument and returns a string representation of the number in octal format.


# hexadecimal representation of a number
# The hex() function is used to convert an integer number to a hexadecimal string prefixed with "0x". the hex() function takes an integer as an argument and returns a string representation of the number in hexadecimal format.

```python
x = 50
print(bin(x))

x = 50
print(oct(x))

x = 50
print(hex(x))
```
