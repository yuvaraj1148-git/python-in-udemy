# base conversion functions
# Python provides built-in functions to convert integers between different bases:
# - bin(): converts an integer to its binary representation (as a string with '0b' prefix)
# - oct(): converts an integer to its octal representation (as a string with '0o' prefix)
# - hex(): converts an integer to its hexadecimal representation (as a string with '0x' prefix)
# - int(): converts a string or float to an integer, with an optional base parameter
# in this function, you give the input in integer and the base you want to convert it to, and it will return the output in string format. The base can be 2 (binary), 8 (octal), 10 (decimal), or 16 (hexadecimal).

```python
# Example usage:
number = 42
print(bin(number))  # Output: 0b101010
print(oct(number))  # Output: 0o52
print(hex(number))  # Output: 0x2a
print(int('1010', 2))  # Output: 10 (binary to decimal)
```

# these function allows boolean values to be converted to integers, where True is converted to 1 and False is converted to 0. This can be useful in various scenarios, such as when performing arithmetic operations or when using boolean values in contexts that require integers.

# also these functions cannot convert non-integer values directly to binary, octal, or hexadecimal. If you try to pass a float or a string that does not represent an integer, it will raise a ValueError. For example:

```python
print(int(True))   # Output: 1
print(int(False))  # Output: 0


try:
    print(bin(3.14))  # This will raise a ValueError
except ValueError as e:
    print(f"Error: {e}")
```
