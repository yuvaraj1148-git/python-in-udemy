```python
# In normal we use like this 
x = 10
y = 5
z = 3

if x>y and y>z:
    print("x is greater than y and y is greater than z")
else:
    print("x is not greater than y or y is not greater than z")

# using changing comparison

x = 10
y = 5
z = 3

if x>y>z:
    print("x is greater than y and y is greater than z")
else:
    print("x is not greater than y or y is not greater than z")

x = 10
y = 10
z = 3

if x == y > z:
    print("x is equal to y and y is greater than z")
else:
    print("x is not equal to y or y is not greater than z")

if x == y == z:
    print("x is equal to y and y is equal to z")
else:
    print("x is not equal to y or y is not equal to z")

x = 11
y = 9
z = 5
a = 90

if x > y > z < a:
    print("x is greater than y, y is greater than z, and z is less than a")
else:
    print("The comparison does not hold true")
```
