# Relational opeartors :
# it compare two vaues and return boolean output
# list of all relational opearators:
# == : equal to
# != : not equal to
# < : less than
# > : greater than
# <= : less than or equal to
# >= : greater than or equal to


```python
#example 
5 == 5  # True
5 != 5  # False
5 < 10  # True
5 > 10  # False
5 <= 10 # True
5 >= 10 # False
```

```python
# if..else statement
if 5 < 10:
    print("5 is less than 10")
else:
    print("5 is not less than 10")
```
