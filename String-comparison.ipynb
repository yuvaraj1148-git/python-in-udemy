```python
# String Comparison in Python
# String comparison are compare using lexicographical order, which is based on the Unicode code points of the characters in the strings.
# 0<1<2 <3.....<< A<B<C......<<< a <b<c...
# In this comparison if the first character of both strings is the same, then the second character is compared, and so on, until a difference is found or the end of one of the strings is reached.
# if a string is a prefix of another string, the shorter string is considered to be less than the longer string.

# Comparing strings
str1 = "Hello"
str2 = "World"
str3 = "Hello"

# Using == operator
print(str1 == str3)  # True
print(str1 == str2)  # False

# Using != operator
print(str1 != str2)  # True
print(str1 != str3)  # False

# Using <, >, <=, >= operators (lexicographical comparison)
print(str1 < str2)   # True
print(str1 > str3)   # False
print(str1 <= str3)  # True
print(str1 >= str3)  # True
```

```python
# Relation on all data types in python
# strings can be compared with all the related operators, and the comparison is based on the lexicographical order of the characters in the strings.
# integer can also be compared with related operators, and the comparison is based on the numerical value of the integers.
# float can also be compared with related operators, and the comparison is based on the numerical value of the floats.
# boolean can also be compared with related operators, and the comparison is based on the truth value of the booleans (True > False).
# complex numbers cannot be compared using relational operators, as they do not have a natural ordering.(it only allow == , !=)
# example of comparing different data types
int1 = 10
int2 = 20
float1 = 10.5
float2 = 20.5
string1 = "apple"
string2 = "banana"
bool1 = True
bool2 = False

# Comparing integers
print(int1 < int2)  # True
print(int1 > int2)  # False
print(int1 == int2)  # False
print(int1 != int2)  # True

# Comparing floats
print(float1 < float2)  # True
print(float1 > float2)  # False
print(float1 == float2)  # False
print(float1 != float2)  # True

# Comparing strings
print(string1 < string2)  # True
print(string1 > string2)  # False
print(string1 == string2)  # False
print(string1 != string2)  # True

# Comparing booleans
print(bool1 < bool2)  # False
print(bool1 > bool2)  # True
print(bool1 == bool2)  # False
print(bool1 != bool2)  # True

# Comparing complex numbers (will raise TypeError)
try:
    complex1 = 1 + 2j
    complex2 = 2 + 3j
    print(complex1 < complex2)  # This will raise TypeError
except TypeError as e:
    print(f"Error: {e}")  # Output: Error: '<' not supported between instances of 'complex' and 'complex'

```
