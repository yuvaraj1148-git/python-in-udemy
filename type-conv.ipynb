# Type Conversion Functions in Python
# This script demonstrates various type conversion functions in Python.
# the script includes examples of converting between different data types such as integers, floats, strings, and booleans.
# in addition, it showcases the use of built-in functions like int(), float(), str(), and bool() for type conversion.
# int() - Converts a value to an integer.
# float() - Converts a value to a floating-point number.
# str() - Converts a value to a string.
# bool() - Converts a value to a boolean



# the int() function can convert a string representation of an integer to an actual integer, but it will raise a ValueError if the string is not a valid integer.

# while converting a string that represents a binary number to an integer, you can use the int() function with base 2.

```python
a = 10
b = 3.14
c = "123"
d = True

print(int(a))     # Output: 10
print(int(b))     # Output: 3
print(int(c))     # Output: 123
print(int(d))     # Output: 1 (True is converted to 1)
print(int('0b1010', 2))  # Output: 10 (binary string to integer)

print(float(a))   # Output: 10.0
print(float(b))   # Output: 3.14
print(float(c))   # Output: 123.0
print(float(d))   # Output: 1.0 (True is converted to 1.0)

print(str(a))     # Output: "10"
print(str(b))     # Output: "3.14"
print(str(c))     # Output: "123"
print(str(d))     # Output: "True"

print(bool(a))    # Output: True
print(bool(b))    # Output: True
print(bool(c))    # Output: True
print(bool(d))    # Output: True
```
