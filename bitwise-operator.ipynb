```python
# format(a,'b') used to see the binary representation of a number
a = 10
print(format(a, 'b'))  # Output: 1010

# also use bin() function to see the binary representation of a number
print(bin(a))  # Output: 0b1010

# the difference between format() and bin() is that format() returns only the binary digits, while bin() returns a string with a '0b' prefix indicating that it is a binary number.

# bitwise operators are used to perform operations on binary numbers. The most common bitwise operators are AND (&), OR (|), XOR (^), NOT (~), left shift (<<), and right shift (>>).
# bitwise AND (&) operator compares each bit of two numbers and returns 1 if both bits are 1, otherwise it returns 0.
b = 6  # binary representation: 0110
print(format(b, 'b'))  # Output: 110
c = 12  # binary representation: 1100
print(format(c, 'b'))  # Output: 1100
print(format(a & b, 'b'))  # Output: 10 # multiplication
print(format(a | b, 'b'))  # Output: 1110 # addition
print(format(a ^ b, 'b'))  # Output: 1100 # like or gate it returns 1 if one of the bits is 1 but not both

# bitwise NOT (~) operator inverts the bits of a number. It changes 1s to 0s and 0s to 1s.
print(format(~a, 'b'))  # Output: -1011 (in two's complement representation, the result is negative)

# left shift (<<) operator shifts the bits of a number to the left by a specified number of positions. Each left shift operation multiplies the number by 2.
print(format(a << 1, 'b'))  # Output: 10100 (10 * 2 = 20) # in left shift the bits are shifted to the left and the empty positions on the right are filled with 0s. The leftmost bit is discarded if it exceeds the size of the integer.
# how many you shift the bits it will multiply the number by 2 for each shift. For example, shifting left by 2 positions will multiply the number by 4 (2^2), shifting left by 3 positions will multiply the number by 8 (2^3), and so on.

# right shift (>>) operator shifts the bits of a number to the right by a specified number of positions. Each right shift operation divides the number by 2.
print(format(a >> 1, 'b'))  # Output: 101 (10 / 2 = 5) # in right shift the bits are shifted to the right and the empty positions on the left are filled with 0s for positive numbers and 1s for negative numbers. The rightmost bit is discarded if it exceeds the size of the integer.
# how many you shift the bits it will divide the number by 2 for each shift. For example, shifting right by 2 positions will divide the number by 4 (2^2), shifting right by 3 positions will divide the number by 8 (2^3), and so on.

# In summary, bitwise operators allow you to manipulate individual bits of integers, which can be useful in various applications such as low-level programming, cryptography, and performance optimization.
# bit_length() method returns the number of bits necessary to represent an integer in binary, excluding the sign and leading zeros.
a = 10
print(a.bit_length())  # Output: 4 (10 in binary is 1010, which requires 4 bits)
```
