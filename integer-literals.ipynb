# integer literals
# in Python, integer literals are used to represent whole numbers. They can be positive or negative and can be expressed in different bases such as decimal (base 10), binary (base 2), octal (base 8), and hexadecimal (base 16).
# it cannot allows floating-point numbers or decimal points. If you try to use a decimal point in an integer literal, it will be treated as a floating-point number instead.
# it can allow to use it in complex numbers, where the integer part is combined with an imaginary part. For example, 3 + 4j represents a complex number with an integer real part of 3 and an imaginary part of 4.

```python
x =5
y = -10
z = 0
print(x)
print(y)
print(z)
```


# decimal literals
# Decimal literals are the most common type of integer literals and are represented using the digits 0 through 9. They can be positive or negative and do not have any prefix.

```python
a = 42
print(a)
```


# binary literals
# Binary literals are represented using the prefix '0b' or '0B' followed by a sequence of binary digits (0s and 1s). They are used to represent numbers in base 2.
# you can use binary literals to represent numbers in a more compact form, especially when dealing with low-level programming or bit manipulation.
# you can calculate the decimal equivalent of a binary literal by using the built-in int() function with base 2.
# you can also calculate it by manually from decimal to binary conversion, which involves dividing the decimal number by 2 and recording the remainders until the quotient is 0. The binary representation is then obtained by reading the remainders in reverse order.

```python
b = 0b1010  # binary representation of 10
print(b)

b = 50
print(bin(b))
```

# octal literals
# Octal literals are represented using the prefix '0o' or '0O' followed by a sequence of octal digits (0 to 7). They are used to represent numbers in base 8.
# you can calculate the decimal equivalent of an octal literal by using the built-in int() function with base 8.
# you can also calculate it by manually from decimal to octal conversion, which involves dividing the decimal number by 8 and recording the remainders until the quotient is 0. The octal representation is then obtained by reading the remainders in reverse order.

```python
c = 0o12  # octal representation of 10
print(c)

c = 50
print(oct(c))
```

# hexadecimal literals
# Hexadecimal literals are represented using the prefix '0x' or '0X' followed by a sequence of hexadecimal digits (0 to 9 and A to F). They are used to represent numbers in base 16.
# you can calculate the decimal equivalent of a hexadecimal literal by using the built-in int() function with base 16.
# you can also calculate it by manually from decimal to hexadecimal conversion, which involves dividing the decimal number by 16 and recording the remainders until the quotient is 0. The hexadecimal representation is then obtained by reading the remainders in reverse order.

```python
d = 0xA  # hexadecimal representation of 10
print(d)

d = 50
print(hex(d))
```
