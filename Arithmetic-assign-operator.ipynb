# Arithmetic Assignment Operator
# The arithmetic assignment operators are used to perform arithmetic operations on a variable and assign the result back to that variable.
# Arithmetic assignment operators combine an arithmetic operation with assignment. The most common arithmetic assignment operators are:
# += : Addition assignment
# -= : Subtraction assignment
# *= : Multiplication assignment
# /= : Division assignment

```python

x = 10
x += 5  # equivalent to x = x + 5
print(x)  # Output: 15

x -= 3  # equivalent to x = x - 3
print(x)  # Output: 12

x *= 2  # equivalent to x = x * 2
print(x)  # Output: 24

x /= 4  # equivalent to x = x / 4
print(x)  # Output: 6.0
```

```python
# Concatenation with strings
greeting = "Hello"
name = "Alice"
message = greeting + ", " + name + "!"
print(message)  # Output: Hello, Alice!
```

```python
# Repetition with strings
a = 'Alice'
b = a * 4
print(b)
```
