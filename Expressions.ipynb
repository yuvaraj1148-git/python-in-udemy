# Expressions in Python
# An expression is a combination of values, variables, operators, and calls to functions that are interpreted and computed by the Python interpreter.
# Examples of expressions:

```python
# Examples of expressions:
5 + 3  # Addition
10 - 4  # Subtraction
7 * 6  # Multiplication
15 // 4  # Floor Division
15 / 4  # Float Division
15 % 4  # Modulus
2 ** 3  # Exponentiation
```

# Precedence of operators:
# precedence means order of operations. The order of operations is as follows:

# 1. Parentheses
# 2. Exponentiation
# 3. Multiplication, Division, Floor Division, Modulus
# 4. Addition, Subtraction

# precedence = 5 + 3 * 2  # Multiplication has higher precedence than addition

# precedence Table:
# precedence matters in expressions. The order of operations is as follows:

# Operator	Precedence
# ()		1
# **		2
# *, /, //, %	3
# +, -		4

```python
# example 

x = 2 + 5 * 10
x = 2 + 50 # here multiplication as higher precedence than addition so it is evaluated first
x = 52 
```

```python
# example 
x = (2 + 5) * 10 # here paranthesis have higher precedence than multiplication so it is evaluated first
x = 7 * 10
x = 70
```

# Associativity of operators:
# In Python, most operators have left-to-right associativity, meaning that when multiple operators of the same precedence level appear in an expression, they are evaluated from left to right. For example:
result = 10 - 5 + 2  # Evaluated as (10 - 5) + 2 = 7

# Associativity list :
# 1. right to left : exponentiation, parenthesis, function call, subscription, attribute reference
# 2. left to right : all other operators(+, -, *, /, //, %, etc.)

```python
# associativity example
x = 2 ** 3 ** 2 # here exponentiation is right associative so it is evaluated from right to left
x = 2 ** 9
x = 512
```

# associativity and precedence means the order in which operations are performed in an expression
# Precedence determines which operator is evaluated first
# Associativity determines the order of evaluation when operators have the same precedence
# Associativity can be left-to-right or right-to-left, depending on the operator. For example, addition and subtraction are left-to-right associative, while exponentiation is right-to-left associative.
# Parentheses can be used to change the order of evaluation.
# Associativity and precedence are important concepts in programming and mathematics, as they can affect the outcome of an expression. Understanding these concepts can help you write more efficient and effective code.

```python
# Associativity and precedence examples:
x = 2 ** 3 ** 2 # here exponentiation is right associative so it is evaluated from right to left
x = 2 ** 9
x = 512
```

# important of parenthesis in expressions

```python
# important of parenthesis in expressions.

x = 2 + 3 * 4 # here multiplication has higher precedence than addition so it is evaluated first
x = 2 + 12
x = 14

# same expression with parenthesis to change the order of evaluation
x = (2 + 3) * 4 # here parentheses change the order of evaluation
x = 5 * 4
x = 20
```
